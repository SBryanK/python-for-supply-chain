## Comparisons and if rules

The numbers 100 and 12 are teaching numbers, not business rules.

```python
quantity = order_line["quantity"]

if quantity >= 100:
    print("Big order line")
elif quantity >= 12:
    print("Medium order line")
else:
    print("Small order line")
```

```python
quantity = order_line["quantity"]

if quantity >= 100:
    return "Big order line".upper()
elif quantity >= 12:
    return "Medium order line".upper()
else:
    return "Small order line".upper()
```
